## Derive Columns

```python
from pyspark.sql.functions import lit
```

```python
df.withColumn("Category",lit("Not Defined")).show()
```

```python
from pyspark.sql.functions import when
 
result_df = df.withColumn(
    "category",
    when(col("exp") >= 10, "Senior")
    .when(col("exp") >= 5, "Mid Level")
    .when(col("exp") > 0, "Junior")
    .otherwise("Invalid Experience"),
).select("exp","category")

display(result_df)
```

```python
result_df.groupBy("category").count().sort("count", ascending=False).display()
```

```python
from pyspark.sql.functions import min, max, avg
result_df.groupBy("category").agg(min("exp"),
                                  max("exp"),
                                  avg("exp")).show()
```
