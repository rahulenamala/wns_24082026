## Create DataFrame from Parquet

```python
df = spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.parquet")
display(df)
```

## Attach Custom Schema

```python
from pyspark.sql.types import *

USER_SCHEMA = StructType([
    StructField("id", IntegerType(), True),
    StructField("age", IntegerType(), True),
    StructField("gender", StringType(), True),
    StructField("designation", StringType(), True),
    StructField("salary", IntegerType(), True)
])
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    schema=USER_SCHEMA
)
display(df)
```

## Handling Bad Records

- PERMISSIVE - (DEFAULT)
- DROPMALFORMED
- FAILFAST

```python
spark.read.option("ColumnNameOfCorruptRecord", "Bad_record").json(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
    mode="PERMISSIVE",
).display()
```

```python
spark.read.option("ColumnNameOfCorruptRecord", "Bad_record").json(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
    mode="DROPMALFORMED",
).display()
```

```python
spark.read.option("ColumnNameOfCorruptRecord", "Bad_record").json(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
    mode="FAILFAST",
).display()
```
